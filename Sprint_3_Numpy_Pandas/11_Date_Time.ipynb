<h2>Datetime</h2>

Datetime represents a date and time as a Python/Pandas object that can be used for calculations and analysis.

A datetime can contain:

<ul><li>Year</li>
<li>Month</li>
<li>Day</li>
<li>Hour</li>
<li>Minute</li>
<li>Second</li></ul>

For example:

2026-08-10 14:30:00

means:

<ul><li>Year - 2026</li>
<li>Day - 10</li>
<li>Hour - 14</li>
<li>Minute - 30</li>
<li>Second - 00</li></ul>

```python
current_datetime = datetime.now()

print(current_datetime)
```

```python
# Create a Specific Datetime

dt = datetime(2026, 10, 8, 14, 30, 0)

print(dt)
```

```python
# Convert Pandas Column to Datetime

orders["Order_Date"] = pd.to_datetime(
    orders["Order_Date"]
)

orders["Delivery_Date"] = pd.to_datetime(
    orders["Delivery_Date"]
)

print(orders.dtypes)
```

<h2>Date Parsing</h2>

Date parsing means converting date information stored as text into a proper datetime format.

For example:

"15-01-2026" is initially a string.

We can convert it using:

pd.to_datetime()

```python
date_value = pd.to_datetime("10-09-2026", dayfirst=True)

print(date_value)
```

```python
# Parsing a DataFrame Column

orders["Order_Date"] = pd.to_datetime(
    orders["Order_Date"]
)
print(orders["Order_Date"])
```

<h2>Timedelta</h2>

A timedelta represents a duration or difference between two dates/times.

__Example:__

January 10 - January 15

The difference is:

5 days

```python
start_date = pd.Timestamp("2026-01-10")
end_date = pd.Timestamp("2026-01-15")

difference = end_date - start_date

print(difference)
```

<h3>Time Zones</h3>

A time zone specifies the region's local time relative to UTC.

For example:

<ul><li>UTC</li>
<li>Asia/Kolkata</li>
<li>America/New_York</li>
<li>Europe/London</li>
<li>Asia/Tokyo</li></ul>

```python
timestamp = pd.Timestamp(
    "2026-10-09 10:00:00",
    tz="UTC"
)

india_time = timestamp.tz_convert(
    "Asia/Kolkata"
)

print(india_time)
```

<h3>Extracting Date Features</h3>

One of the most important uses of datetime data in Data Science is extracting useful information from dates.

Pandas provides the .dt accessor.

```python
# Extract Year
orders["Year"] = orders["Order_Date"].dt.year

print(orders["Year"])
```

```python
# Extract Month
orders["Month"] = orders["Order_Date"].dt.month

print(orders["Month"])
```

```python
# Extract Month Name
orders["Month_Name"] = (
    orders["Order_Date"].dt.month_name()
)

print(orders["Month_Name"])
```

```python
# Extract Day
orders["Day"] = orders["Order_Date"].dt.day
print(orders["Day"])
```

```python
# Extract Day Name
orders["Day_Name"] = (
    orders["Order_Date"].dt.day_name()
)

print(orders["Day_Name"])
```

```python
# Extract Day of Week
orders["Day_of_Week"] = (
    orders["Order_Date"].dt.dayofweek
)

print(orders["Day_of_Week"])
```

```python
# Extract Quarter
orders["Quarter"] = (
    orders["Order_Date"].dt.quarter
)

print(orders["Quarter"])
```

<h3>Extract Multiple Features</h3>

Instead of creating each feature separately, we can create several.

```python
orders["Year"] = orders["Order_Date"].dt.year
orders["Month"] = orders["Order_Date"].dt.month
orders["Day"] = orders["Order_Date"].dt.day
orders["Quarter"] = orders["Order_Date"].dt.quarter
orders["Day_of_Week"] = orders["Order_Date"].dt.dayofweek
orders["Is_Weekend"] = orders["Order_Date"].dt.dayofweek >= 5

print(orders)
```
