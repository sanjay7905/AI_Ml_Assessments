<h2>split()</h2>

split() divides a string into smaller parts based on a separator.

Syntax:

Series.str.split()

```python
# Names have full name
customers["Full_Name"] = [
    "Aishu Devendran",
    "Rahul Kumar",
    "Priya Sharma",
    "John David",
    "Meena Raj"
]

print(customers[["Full_Name"]])
```

```python
# splitting name as full name and last name
customers[["First_Name", "Last_Name"]] = (
    customers["Full_Name"]
    .str.split(" ", expand=True)
)

print(customers[[
    "Full_Name",
    "First_Name",
    "Last_Name"
]])
```

__Example:__

Suppose an employee dataset contains full name we can split it into first name and lastname

<h2>replace()</h2>

String replace() substitutes one piece of text with another.

Syntax:

Series.str.replace(old, new)

```python
cities = pd.Series([
    "Chennai",
    "Madras",
    "Chennai",
    "Madras"
])
cities = cities.str.replace(
    "Madras",
    "Chennai"
)

print(cities)
```

```python
# Remove Special Characters

phone = pd.Series([
    "987-654-3210",
    "912-345-6780"
])

phone = phone.str.replace(
    "-",
    "",
    regex=False
)

print(phone)
```

__Example:__

Customer phone numbers may be stored with hyphens we can clean the formatting before analysis.

<h2>Regular Expressions (Regex)</h2>

```python
import re
```

Regular Expressions (Regex) are patterns used to search, match, extract, replace, or validate text.

Python provides the re module for regular expressions, while Pandas string methods support regex directly.

Regex is especially useful when the text follows a pattern rather than an exact value.

<h3>Regex – Extract Phone Numbers</h3>

```python
data = pd.Series([
    "Contact: 9876543210",
    "Call: 9123456780",
    "Phone: 9988776655"
])

data.str.extract(r"(\d{10})")
```

Therefore, \d{10} means 10 consecutive digits.

<h3>Regex – Remove Special Characters</h3>

```python
names = pd.Series([
    "Arjun@123",
    "Rogan#456",
    "Prakash$789"
])

clean_names = names.str.replace(
    r"[^A-Za-z0-9]",
    "",
    regex=True
)

print(clean_names)
```

[A-Za-z0-9] means letters and numbers.

^ inside brackets means NOT.

So, [^A-Za-z0-9] means:

Any character that is not a letter or number.

<h3>Regex – Find Words Containing a Pattern</h3>

```python
products = pd.Series([
    "iPhone",
    "Samsung Phone",
    "Laptop",
    "Phone Case"
])

products.str.contains(
    r"phone",
    case=False,
    regex=True
)
```
