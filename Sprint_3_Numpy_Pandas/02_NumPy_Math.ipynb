<h1>NumPy Mathematical Operations</h1>


```python
import numpy as np
scores = np.array([78, 85, 92, 67, 88, 95, 73, 81])

print(scores)
```

<h2>Sum</h2>

The Sum function calculates the total of all elements in an array.

It is useful when finding the total marks, total sales, total revenue, or total values in a dataset.

```python
total = np.sum(scores)

print(total)
```

<h3>AI/ML use</h3>

Sum can be used when calculating:

Total sales<br>
Total revenue<br>
Total number of records

<h2>Mean</h2>

The Mean is the average value of all numbers.

It is calculated by dividing the sum of all values by the total number of values.

```python
average = np.mean(scores)

print(average)
```

<h3>AI/ML use</h3>

Mean is commonly used for:

Understanding the average value of a feature<br>
Missing-value replacement<br>
Data analysis<br>
Feature scaling

<h2>Median</h2>

The Median is the middle value after arranging the numbers in ascending order.

If there are two middle values, NumPy returns their average.

```python
median = np.median(scores)

print(median)
```

<h3>AI/ML use</h3>

Median is useful for:

Handling outliers<br>
Missing-value imputation<br>
Understanding data distribution

<h2>Variance</h2>

Variance measures how far the data points are spread from the mean.

Small variance - values are close together.
    
Large variance - values are widely spread.

```python
variance = np.var(scores)

print(variance)
```

<h3>AI/ML use</h3>

Variance is important for:

Understanding data spread<br>
Feature selection<br>
Detecting low-variance features

<h2>Standard Deviation</h2>

Standard Deviation measures the average distance of data points from the mean.

It is simply the square root of the variance.

```python
std = np.std(scores)

print(std)
```

<h3>AI/ML use</h3>

Standard deviation is heavily used in:

Feature scaling<br>
StandardScaler<br>
Outlier detection

<h2>Minimum</h2>
Minimum means finding the smallest value in a dataset.

```python
minimum = np.min(scores)

print(minimum)
```

<h3>AI/ML use</h3>

Minimum can be used to:

Find the lowest price<br>
Find minimum temperature<br>
Find minimum model error

<h2>Maximum</h2>


Maximum means finding the largest value in a dataset.

```python
maximum = np.max(scores)

print(maximum)
```

<h3>AI/ML use</h3>

Maximum can be used to:

Find highest salary<br>
Find highest score<br>
Find maximum temperature

<h2>Argmin</h2>

argmin() does NOT return the minimum value.

It returns the index/location of the minimum value.

```python
print(np.argmin(scores))
```

Because

scores[3] = 67

<h2>Argmax</h2>

argmax() returns the index/location of the largest value.

```python
print(np.argmax(scores))
```

Because

scores[5] = 95

<h2>Percentile</h2>

A percentile tells us the value below which a certain percentage of data falls.

Example:

50th percentile = Median

If you are at the 90th percentile in an exam, it means your score is higher than or equal to roughly 90% of the observations in the reference group.

25th percentile<br>
50th percentile (Median)<br>
75th percentile

```python
print(np.percentile(scores,25))

print(np.percentile(scores,50))

print(np.percentile(scores,75))
```

<h3>AI/ML use</h3>

Percentiles are useful for:

Understanding data distribution<br>
Detecting outliers<br>
Finding Q1 and Q3

<h2>Dot Product</h2>

The dot product multiplies corresponding elements of two vectors and then adds the results.

It is one of the most important operations in Machine Learning.

```python
a = np.array([2,3,4])

b = np.array([5,6,7])

print(np.dot(a,b))
```

<h2>Matrix Multiplication</h2>

Matrix multiplication combines two matrices to produce another matrix.

The number of columns in the first matrix must equal the number of rows in the second matrix.

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])
print(np.matmul(A,B))
```
