<h2>What is NumPy?</h2>

NumPy stands for Numerical Python.

It is a Python library mainly used for working with numbers, arrays, mathematical calculations, and data. It provides a powerful object called ndarray (N-dimensional array) that stores data in a compact and optimized format.

<h2>Dimensions</h2>

Dimension tells us how many levels/axes an array has.

```python
[1,2,3,4] - 1D
[[1,2],[3,4] - 2D
[[[1]]] - 3D
```

```python
import numpy as np

a = np.array([1,2,3,4])

b = np.array([[1,2],[3,4]])

c = np.array([[[1],[2]],[[3],[4]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
```

<h2>Copy vs View</h2>

<h4>Copy</h4>

Creates a new independent array.

Changing the copy does not change the original.

```python
a = np.array([10, 20, 30])

b = a.copy()

b[0] = 100
print(a)
print(b)
```

<h4>View</h4>

Creates another way of looking at the same underlying data.

Changing the view can change the original.

```python
a = np.array([10,20,30])

b = a.view()

b[0] = 100

print(a)
print(b)
```

<h2>AI/ML Usage</h2>

Copies protect original datasets while preprocessing.

Views save memory for large datasets.

<h2>Broadcasting</h2>
Broadcasting allows NumPy to perform operations between arrays of different shapes when their shapes are compatible.

```python
import numpy as np

salary = np.array([20000,22000,30000])

new_salary = salary + 5000

print(new_salary)

matrix = np.array([[1,2,3],
                   [4,5,6]])

vector = np.array([10,20,30])

print(matrix + vector)
```

<h3>AI/ML Usage</h3>

Feature scaling

New Value = Original Value - Mean

Mean values are broadcast across every row.

<h2>Array Operations</h2>

Array operations mean performing mathematical operations directly on NumPy arrays.

NumPy can perform an operation on all elements of an array at once.

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)

print(a-b)

print(a*b)

print(a/b)
```

```python
print(a > 2)
print(a[a > 2])
```

Aggregation Operations

NumPy also provides functions to summarize an array.

```python
print(np.sum(a))

print(np.mean(a))

print(np.max(a))

print(np.min(a))
```

<h2>Universal Functions (ufuncs)</h2>

Universal functions, commonly called ufuncs, are NumPy functions that operate on array elements efficiently.

A ufunc applies a mathematical operation to each element of a NumPy array.

```python
#square root
arr = np.array([1,4,9,16])

print(np.sqrt(arr))
```

```python
arr = np.array([1,2,3])

print(np.square(arr))

#Exponential – np.exp()

print(np.exp(arr))

print(np.log(arr))

print(np.sin(arr))

print(np.cos(arr))

#Absolute Value – np.abs()

print(np.abs([-2,-5,6]))
```
