<h2>Create a NumPy array</h2>

A NumPy array is a data structure used to store numerical data and perform mathematical operations efficiently.

Real-world example: Store the marks of five students and calculate their average.

```python
import numpy as np

marks = np.array([80, 75, 90, 65, 85])

print(marks)
print("Shape:", marks.shape)
print("Average:", np.mean(marks))
```

<h2>Difference between a NumPy array and a Python list</h2>

A Python list can store different data types, whereas a NumPy array usually stores elements of the same data type and supports efficient vectorized mathematical operations.

```python
import numpy as np

# Python list
numbers_list = [10, 20, 30]
print(numbers_list * 2)

# NumPy array
numbers_array = np.array([10, 20, 30])
print(numbers_array * 2)
```

<h2>Explain broadcasting with an example</h2>

Broadcasting allows NumPy to perform operations between arrays of compatible shapes without requiring you to manually repeat the smaller array.
    
Real-world example: Increase each student's mark by 5 bonus points.

```python
import numpy as np

marks = np.array([60, 70, 80])
bonus = 5

result = marks + bonus

print(result)
```

<h2>Difference between Series and DataFrame</h2>

Both are Pandas data structures used to work with data.
    
Series: One-dimensional, like a single column.
    
DataFrame: Two-dimensional, like a complete table with rows and columns.

```python
import pandas as pd

# Series
marks = pd.Series([80, 90, 75], name="Marks")
print(marks)

# DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi"],
    "Marks": [80, 90, 75]
})

print(df)
```

<h2>Read a CSV file using Pandas</h2>

Definition: A CSV file stores tabular data as comma-separated values. Pandas reads it into a DataFrame using read_csv().

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [22, 23, 21],
    "Salary": [25000, 30000, 28000]
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
```

```python
df = pd.read_csv("employees.csv")

print(df)
print(df.head())
```

<h2>Merge two DataFrames</h2>

Merging combines two DataFrames using a common column, similar to joining tables in SQL.
    
Real-world example: Combine employee details with department names using DeptID

```python
import pandas as pd

employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Arun", "Priya", "Ravi"],
    "DeptID": [101, 102, 101]
})

departments = pd.DataFrame({
    "DeptID": [101, 102],
    "Department": ["IT", "HR"]
})

result = pd.merge(
    employees,
    departments,
    on="DeptID",
    how="inner"
)

print(result)
```

<h2>Perform GroupBy and aggregation</h2>

groupby() divides rows into groups, and aggregation calculates summary values for each group.
    
Example: Calculate the average salary and total salary for each department.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [30000, 25000, 40000, 35000, 50000]
})

result = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Total_Salary=("Salary", "sum"),
    Employee_Count=("Salary", "count")
)

print(result)
```

<h2>Filter rows based on multiple conditions</h2>

Boolean filtering selects rows that satisfy specified conditions.
    
Example: Select employees whose salary is above ₹25,000 and whose age is at least 23.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Age": [22, 24, 25, 23],
    "Salary": [30000, 35000, 24000, 40000]
})

result = df[
    (df["Salary"] > 25000) &
    (df["Age"] >= 23)
]

print(result)
```

<h2>Handle missing values</h2>

Missing values represent data that is unavailable or unrecorded. Pandas commonly represents them as NaN.
    
Example: Fill missing ages with the average age and remove rows with missing names.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Arun", "Priya", None, "Ravi"],
    "Age": [22, np.nan, 24, 26],
    "Salary": [25000, 30000, np.nan, 28000]
})

print("Missing values:")
print(df.isnull().sum())

# Fill missing numerical values with column means
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Remove rows where Name is missing
df = df.dropna(subset=["Name"])

print(df)
```

<h2>Calculate statistical measures using NumPy</h2>

Definition: Statistical measures summarize numerical data, helping us understand its center and variability.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("25th Percentile:", np.percentile(data, 25))
```

<h2>Optimize a DataFrame for memory usage</h2>
    
Memory optimization reduces the amount of RAM used by a DataFrame, especially when it contains many rows or unnecessarily large data types.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 30, 28, 35],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [25000, 30000, 45000, 32000, 50000]
})

print("Before optimization:")
print(df.memory_usage(deep=True))
print("Total bytes:", df.memory_usage(deep=True).sum())

# Reduce numerical data types safely for this sample
df["Age"] = df["Age"].astype("int8")
df["Salary"] = pd.to_numeric(df["Salary"], downcast="integer")

# Store repeated labels efficiently
df["Department"] = df["Department"].astype("category")

print("\nAfter optimization:")
print(df.memory_usage(deep=True))
print("Total bytes:", df.memory_usage(deep=True).sum())
```
