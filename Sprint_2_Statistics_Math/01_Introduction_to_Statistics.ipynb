<h1>01_Introduction_to_Statistics</h1>

<h2>What is Statistics?</h2>

Statistics is the branch of mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data to make better decisions.
    
Simply put, statistics helps us understand data and find useful information from it.

Example:

A hospital wants to know the average age of patients visiting every day. It collects patient ages, calculates the average, and uses the result to plan healthcare services.

<h2>Why Statistics is Important in AI/ML</h2>

Statistics is the foundation of Artificial Intelligence and Machine Learning because AI models learn patterns from data. Statistics helps understand data before building models.

Importance:

<ul><li>Understand data distribution</li>

<li>Handle missing values</li>

<li>Detect outliers</li>

<li>Select important features</li>

<li>Evaluate model performance</li>

<li>Make predictions</li></ul>

Example:

A bank wants to predict whether a customer will repay a loan.

Statistics helps to:

<uL><li>Analyze customer income</li>
    
<li>Calculate average salary</li>
    
<li>Find risky customers</li>
    
<li>Understand relationships between variables</il></ul>

Machine Learning then uses this information to predict loan approval.

<h2>Statistics vs Machine Learning</h2>

<ul><li>Statistics focuses on understanding data meanwhile Machine Learning focuses on making predictions.</li>

<li>Statistics explains relationships but Machine Learning learns patterns automatically</li>

<li>Statistics uses mathematical models while Machine Learning uses algorithms</li>

<li>Statistics smaller datasets are sufficient meanwhile Machine Learning usually requires larger datasets</li></ul>

Example:

Statistics: A teacher calculates the average marks of students.

Machine Learning: A system predicts whether a student will pass based on previous records.

<h2>Descriptive Statistics</h2>

Descriptive statistics summarizes and describes data using numbers and graphs.

<h3>Central Tendency</h3>

Central tendency tells us where the center or typical value of the data is.

Mean: The average of all the values. We add all the values and divide by the number of values.

Example: 10, 20, 30 - Mean = 20

Median: The middle value when the data is arranged from smallest to largest.

Example: 10, 20, 30 - Median = 20

Mode: The value that appears most frequently in the data.

Example: 10, 20, 20, 30 - Mode = 20

<h3>Variability (Spread)</h3>

Variability tells us how much the data values differ from each other or from the center.

Range: The difference between the largest and small

Variance: Shows how far the data values are spread out from the mean.

Standard Deviation: It is the square root of variance.

<h3>Frequency Distribution</h3>

A frequency distribution shows how many times each value or category appears in the data.

Counts and Percentages

Count: Simply tells us how many times something occurs.

Percentage: Tells us what portion of the total that count represents, usually out of 100.


```python
import pandas as pd

marks = pd.Series([80, 75, 90, 95, 88])

print(marks.describe())
```

<h2>Inferential Statistics</h2>

Inferential statistics uses a sample to make conclusions about an entire population.

Instead of studying everyone, we study a small group.

Example:

You taste one spoonful of soup to know if the whole pot needs more salt.

<h2>population</h2>

A population means the complete group of people, objects, or data that we want to study or understand.

Example:
If we want to know the average height of students in a college, all students in that college are the population.

Easy tips: Population = The whole group.

<h2>sample</h2>

A sample is a small part of a population that we choose to study.

Example:
If a college has 1,000 students, asking all 1,000 students about their favorite food may take a lot of time. Instead, we ask 100 students.

Sample = A smaller part taken from the whole population.

<h2>census</h2>

A census means collecting information from every member of the population.

Example:
If a college has 1,000 students and we collect information from all 1,000 students, it is a census.

Census = Study the whole population.

<h2>Sampling</h2>

Sampling is the process of selecting a subset of the population for study.

Example:

A food company tastes products using 100 customers before launching them nationwide

Sampling = Choosing a sample from the population.


```python
import pandas as pd

data = pd.DataFrame({
    "Age":[21,22,23,24,25,26,27,28]
})

sample = data.sample(3)

print(sample)
```
