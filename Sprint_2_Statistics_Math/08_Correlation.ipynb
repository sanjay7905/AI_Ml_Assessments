<h1>Correlation & Covariance</h1>

<h2>Correlation</h2>

Correlation measures the strength and direction of the relationship between two variables.

Its value ranges from -1 to +1.

+1 - Perfect Positive Correlation

-1 - Perfect Negative Correlation

0 - No Correlation

Example:

The school wants to know whether students who study more hours score higher marks.

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [45, 50, 55, 65, 70, 75, 85, 90]
})

print(data)
```

```python
print(data.corr())
```

A high positive correlation means that as study hours increase, marks also increase.

<h2>Covariance</h2>

Covariance measures how two variables change together.

Positive Covariance - Variables increase together. Negative Covariance - One increases while the other decreases.

Unlike correlation, covariance does not have a fixed range.

Example:

Study hours and marks both increase together.

```python
print(data.cov())
```

A positive covariance indicates that study hours and marks move in the same direction.

<h2>Positive Correlation</h2>

A Positive Correlation means that when one variable increases, the other variable also increases.

Example:

Students who study more generally score higher marks.

```python
print(data["Study_Hours"].corr(data["Marks"]))
```

More study hours - Higher marks.

<h2>Negative Correlation</h2>

A Negative Correlation means that when one variable increases, the other decreases.

Example:

Suppose the school also records the number of hours spent watching TV.

Students who watch more TV generally score lower marks.

```python
import pandas as pd

data2 = pd.DataFrame({
    "TV_Hours":[1,2,3,4,5,6,7,8],
    "Marks":[90,85,80,75,70,65,60,55]
})

print(data2.corr())
```

More TV hours - Lower marks.

<h2>No Correlation</h2>

No Correlation means there is no relationship between two variables.

Example:

A student's roll number has no relationship with exam marks.

```python
import pandas as pd

data3 = pd.DataFrame({
    "Roll_No":[1,2,3,4,5,6,7,8],
    "Marks":[75,60,85,55,90,70,65,80]
})

print(data3.corr())
```

Changing the roll number does not affect marks.

<h2>Pearson Correlation</h2>

Pearson Correlation measures the strength of a linear relationship between two numerical variables.

The value ranges from -1 to +1.

Example:

The school measures the relationship between study hours and marks.

```python
pearson = data["Study_Hours"].corr(data["Marks"], method="pearson")

print(pearson)
```

A value close to +1 indicates a strong positive linear relationship.

<h2>Spearman Correlation</h2>

Spearman Correlation measures the relationship between the rank (order) of two variables.

It is useful when the relationship is not perfectly linear.

Example:

The school ranks students based on study hours and exam marks.

```python
spearman = data["Study_Hours"].corr(data["Marks"], method="spearman")

print(spearman)
```

A value close to +1 means that students with higher study-hour ranks also have higher mark ranks.

<h2>Correlation Matrix</h2>

A Correlation Matrix is a table that shows the correlation between all numerical variables in a dataset.

Example:

The school wants to compare multiple variables, such as study hours, attendance, and marks.

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours":[2,3,4,5,6,7,8,9],
    "Attendance":[60,65,70,75,80,85,90,95],
    "Marks":[45,50,55,65,70,75,85,90]
})

print(data.corr())
```

The matrix helps identify which variables are strongly related.

<h2>Heatmap</h2>

A Heatmap is a graphical representation of the correlation matrix using colors.

Darker colors represent stronger relationships.
    
Lighter colors represent weaker relationships.
    
Example:

The school visualizes the relationships between study hours, attendance, and marks.

```python
import matplotlib.pyplot as plt

correlation = data.corr()

plt.imshow(correlation, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")
plt.show()
```

The heatmap makes it easy to identify variables that have strong positive or negative correlations.
