<h1>Hypothesis Testing</h1>

<h2>What is Hypothesis Testing?</h2>

Hypothesis Testing is a statistical method used to determine whether there is enough evidence to accept or reject a claim about a population.

```python
# sample dataset

sample = [9.5, 10.2, 9.8, 10.1, 9.7, 10.3, 9.9, 10.0, 9.6, 10.1]
```

<h2>Null Hypothesis (H₀)</h2>

The Null Hypothesis (H₀) is the default assumption that there is no significant difference or effect.

Example:

The company claims:

Average battery life = 10 hours

So,

H₀ : μ = 10 hours

```python
claim = 10

print("Null Hypothesis: Average battery life =", claim, "hours")
```

<h2>Alternative Hypothesis (H₁)</h2>

The Alternative Hypothesis (H₁) states that there is a significant difference or effect.

Example:

Suppose we believe the average battery life is not equal to 10 hours.

H₁ : μ ≠ 10 hours

```python
print("Alternative Hypothesis: Average battery life is not 10 hours")
```

<h2>p-value</h2>

The p-value is the probability of obtaining the observed result if the null hypothesis is true.

Decision:

p-value ≤ 0.05 - Reject H₀

p-value > 0.05 - Fail to reject H₀

Example:

Suppose, p-value = 0.03

Since, 0.03 < 0.05

Reject the null hypothesis.

```python
p_value = 0.03

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

<h2>Confidence Interval</h2>

A Confidence Interval (CI) is a range of values that is likely to contain the true population parameter.

A 95% confidence interval means we are 95% confident that the true population mean lies within the calculated interval.

Example:

Battery life confidence interval:

9.7 hours to 10.2 hours

```python
import numpy as np

sample = [9.5, 10.2, 9.8, 10.1, 9.7, 10.3, 9.9, 10.0, 9.6, 10.1]

print("Mean:", np.mean(sample))
```

<h2>Significance Level (α)</h2>

The Significance Level (α) is the maximum probability of making a Type I Error.

The most commonly used value is:

α = 0.05

Example:

If α = 0.05

There is a 5% chance of rejecting a true null hypothesis.

```python
alpha = 0.05

print("Significance Level:", alpha)
```

<h2>Type I Error</h2>

A Type I Error occurs when we reject the null hypothesis even though it is actually true.

It is also called a False Positive.

Example:

The battery actually lasts 10 hours, but we incorrectly conclude that it does not.

```python
print("Type I Error: Rejecting a true Null Hypothesis")
```

<h2>Type II Error</h2>

A Type II Error occurs when we fail to reject the null hypothesis even though it is false.

It is also called a False Negative.

Example:

The battery actually lasts 9 hours, but we incorrectly conclude that it still lasts 10 hours.

```python
print("Type II Error: Failing to reject a false Null Hypothesis")
```

<h2>One-tailed Test</h2>

A One-tailed Test checks whether a parameter is greater than or less than a specific value.

Right-tailed Test, H₁ : μ > 10

Left-tailed Test, H₁ : μ < 10

Example:

A company claims the battery lasts more than 10 hours.

```python
sample_mean = 10.2

if sample_mean > 10:
    print("Supports Right-Tailed Test")
```

<h2>Two-tailed Test</h2>

A Two-tailed Test checks whether a parameter is different from a specified value.

H₁ : μ ≠ 10

Example:

We simply want to know whether the battery life is different from 10 hours, regardless of whether it is higher or lower.

```python
sample_mean = 9.7

if sample_mean != 10:
    print("Supports Two-Tailed Test")
```
