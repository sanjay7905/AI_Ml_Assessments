<h1>Data Quality Assessment</h1>

```python
import pandas as pd

df = pd.read_csv("customer_churn.csv")
df.head()
```

<h2><u>Missing Values</u></h2>

A missing value is a value that is not available or has not been recorded for a particular observation.

In Pandas, missing values are commonly represented as: NaN

Example:

Suppose a telecom company's customer data looks like:

  Customer   Age  Monthly Charges
0     C001  25.0             50.5
1     C002   NaN             70.2
2     C003  32.0              NaN

Here, the age of C002 and monthly charge of C003 are missing.

<h4>AI/ML usage</h4>

Missing values are important because many ML algorithms cannot directly work with missing values.

Common solutions include:

- Removing rows
- Removing columns
- Mean/median imputation
- Mode imputation
- Forward/backward filling
- Model-based imputation

<h2>Number of Missing Values</h2>

```python
missing_count = df.isnull().sum()

print(missing_count)
```

<h2>Total Number of Missing Values</h2>

The total number of missing values is the sum of all missing cells across the entire dataset.

```python
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

<h2>Percentage of Missing Values</h2>

Missing-value percentage tells us what percentage of the values in a column are missing.

Formula:

Missing Percentage = (Number of Missing Values /bTotal Number of Rows) ×100

```python
missing_percentage = df.isnull().mean() * 100

print(missing_percentage)
```

<h2>Missing Values by Column</h2>

Missing values by column tells us which columns contain missing data and how much.

```python
missing_by_column = df.isnull().sum()

missing_by_column = missing_by_column[missing_by_column > 0]
print(missing_by_column)
```

<h2>Missing Value Patterns</h2>

A missing-value pattern shows whether missing values occur randomly, together, or in particular groups of records.

For example:

Customer A → Age missing

Customer B → MonthlyCharges missing

Customer C → Age + TotalCharges missing

Understanding patterns can help determine why values are missing.

<h4>AI/ML usage</h4>
    
Missing-value patterns can indicate:

- Data collection problems
- System errors
- Optional fields
- Customer behavior
- Potential bias
    
Before deciding on imputation, we should understand the pattern.

```python
df[df.isnull().any(axis=1)]
```

<h2><u>Duplicate Records</u></h2>

A duplicate record is a row that appears more than once in the dataset.

Example:

A customer might accidentally be entered twice:

| **CustomerID** | **Age** | **Churn** |
| -------------- | ------: | --------- |
| C001           |      25 | No        |
| C001           |      25 | No        |

The second row is a duplicate.

<h4>AI/ML usage</h4>

Duplicates can cause problems because they may:

- Bias statistics
- Overrepresent certain customers
- Affect model training
- Cause data leakage in some situations

<h2>Identify Duplicates</h2>

```python
df.duplicated()
```

True means that the row is a duplicate of an earlier row.

<h2>Count Duplicates</h2>

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

<h2>Identify Completely Duplicated Rows</h2>

Completely duplicated rows are rows in a dataset where all column values are exactly the same as those in an earlier row.

```python
complete_duplicates = df[df.duplicated(keep=False)]

print(complete_duplicates)
```

keep=False marks all copies of the duplicate instead of marking only the second occurrence.

```python
print(
    "Rows involved in complete duplication:",
    df.duplicated(keep=False).sum()
)
```

<h2>Potential Duplicate Records</h2>

A potential duplicate is a record that may represent the same entity even though the entire row is not identical.

For example:

| **CustomerID** | **Age** | **Gender** |
| -------------- | ------: | ---------- |
| C001           |      25 | Female     |
| C001           |      25 | female     |

These aren't completely identical because of capitalization, but they may represent the same customer.

```python
df["CustomerID"].duplicated().sum()
```

<h2>Empty Strings</h2>

An empty string is a value such as:

""

It isn't necessarily recognized by Pandas as NaN.

```python

empty_string_counts = (
    df.select_dtypes(include="str")
      .apply(lambda x: x.str.strip().eq("").sum())
)

print(empty_string_counts)
```

<h2>Null-like Values</h2>

Sometimes missing data is represented using text instead of actual NaN.

Examples:

"NA"

"N/A"

"null"

"None"

"missing"

"-"

"unknown"

These are called null-like values.

```python
null_like_values = [
    "",
    "NA",
    "N/A",
    "NULL",
    "null",
    "None",
    "none",
    "missing",
    "-"
]

for column in df.select_dtypes(include="str"):
    for value in null_like_values:
        count = (
            df[column]
            .astype(str)
            .str.strip()
            .eq(value)
            .sum()
        )
        
        if count > 0:
            print(column, value, count)
```
