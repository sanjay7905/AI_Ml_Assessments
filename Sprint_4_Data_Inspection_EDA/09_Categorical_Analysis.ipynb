# Categorical Data Analysis

### 1. Unique Categories

 The distinct, non-duplicate values present within a categorical feature column.

Example: 

Payment methods in a customer database (Credit Card, Bank Transfer, Electronic Check, Mailed Check).

#### AI/ML Usage

Determines vocabulary size for feature encoding schemes such as One-Hot Encoding or Entity Embeddings.

```python
import pandas as pd

df = pd.read_csv('customer_churn.csv')
unique_payments = df['PaymentMethod'].unique()
num_unique = df['PaymentMethod'].nunique()

print(f"Number of Unique Categories: {num_unique}")
print(f"Categories: {unique_payments}")
```

### 2. Category Frequencies

The absolute raw count of records corresponding to each unique category in a column.

Example: 

Counting retail sales records per shirt size (S: 120, M: 450, L: 300).

#### AI/ML Usage

Helps tree-based algorithms compute split criteria like Gini Impurity or Information Gain.

```python
contract_counts = df['ContractType'].value_counts()
print(contract_counts)
```

### 3. Category Distribution

The relative percentage or proportion of each category relative to the total number of dataset rows.

Example:

Analyzing website traffic by device type (55% Mobile, 40% Desktop, 5% Tablet).

#### AI/ML Usage

Used to verify that train-test splits maintain identical proportions via Stratified K-Fold cross-validation.

```python
contract_distribution = df['ContractType'].value_counts(normalize=True) * 100
print(contract_distribution)
```

### 4. Rare Categories

Categories that appear with very low frequency (typically under 1% to 5% of the total dataset).

Example:

Obsolete operating systems appearing in web application login logs (Windows 98).

#### AI/ML Usage

Causes overfitting and high variance; rare categories are typically grouped into an Other bin prior to modeling.

```python
threshold = 0.05
freq = df['PaymentMethod'].value_counts(normalize=True)
rare_cats = freq[freq < threshold].index

df['PaymentMethod_Grouped'] = df['PaymentMethod'].replace(rare_cats, 'Other')
print(df['PaymentMethod_Grouped'].value_counts())
```

### 5. High Cardinality

A categorical variable containing a very high number of distinct unique values relative to total observations.

Example:

Customer ID, ZIP codes, device IP addresses, or product SKU codes.

#### AI/ML Usage

Causes explosion of feature dimensions (curse of dimensionality) under One-Hot Encoding. Solved using Target Encoding, Frequency Encoding, or Hashing.

```python
cardinality = df['CustomerID'].nunique()
print(f"Customer ID Unique Count: {cardinality}")

# Target Encoding High Cardinality Features

target_map = df.groupby('ContractType')['Churn'].apply(lambda x: (x == 'Yes').mean())
df['Contract_TargetEnc'] = df['ContractType'].map(target_map)
```

### 6. Low Cardinality

A categorical feature containing a small, manageable number of unique categories (e.g., 2 to 10 distinct values).

Example: 

Marital status (Single, Married, Divorced), subscription tier (Free, Pro, Enterprise).

#### AI/ML Usage

Ideal candidate for One-Hot Encoding or Binary Encoding without risk of memory saturation.

```python
low_card_cols = [col for col in df.select_dtypes(include='str').columns if df[col].nunique() <= 5]
print("Low Cardinality Columns:", low_card_cols)
```

### 7. Category Imbalance

A scenario where one or more categories disproportionately dominate the feature or target label distribution.

Example:

Fraud detection datasets containing 99.2% legitimate transactions and 0.8% fraudulent ones.

#### AI/ML Usage

Causes classifiers to bias toward predicting the majority class. Fixed using class weighting, SMOTE, or Focal Loss.

```python
churn_imbalance = df['Churn'].value_counts(normalize=True) * 100
print(churn_imbalance)
```
